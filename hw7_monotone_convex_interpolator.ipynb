### Include an IFR curve plot and briefly explain how your method ensures monotonicity and convexity.

```python
import numpy as np
import matplotlib.pyplot as plt
from fixedincomelib.utilities.numerics import InterpolatorFactory, InterpMethod, ExtrapMethod

import QuantLib as ql

# Same time axis the yield curve uses: Actual/Actual (ISDA) year fraction from the value date.
def to_ql(d):
    ts = pd.Timestamp(d)
    return ql.Date(ts.day, ts.month, ts.year)
isda = ql.ActualActual(ql.ActualActual.ISDA)
t_nodes = np.array([isda.yearFraction(to_ql("2026-01-02"), to_ql(d)) for d in node_dates])
grid = np.linspace(t_nodes[0] - 0.3, t_nodes[-1] + 0.5, 800)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.2), sharex=True)
for ax, (name, q) in zip(axes, [("SOFR-1B", sofr_quotes), ("FF-1B", ff_quotes)]):
    y = q["values"].to_numpy()
    mc = InterpolatorFactory.create_1d_interpolator(t_nodes, y, InterpMethod.MONOTONE_CONVEX, ExtrapMethod.FLAT)
    pcp = InterpolatorFactory.create_1d_interpolator(t_nodes, y, InterpMethod.PIECEWISE_CONSTANT_LEFT_CONTINUOUS, ExtrapMethod.FLAT)
    ax.step(grid, [pcp.interpolate(t) for t in grid], where="post", color="0.65", lw=1.2, label="piecewise constant")
    ax.plot(grid, [mc.interpolate(t) for t in grid], color="C0", lw=2, label="monotone convex")
    ax.plot(t_nodes, y, "o", color="C3", label="IFR nodes")
    ax.set_title(f"{name} instantaneous forward rate")
    ax.set_xlabel("years from value date")
    ax.set_ylabel("IFR")
    ax.grid(alpha=0.3)
axes[0].legend()
plt.tight_layout()
plt.show()

# Numerical check on every node interval: nodes hit, monotone in the direction of the data, convex.
for name, q in [("SOFR-1B", sofr_quotes), ("FF-1B", ff_quotes)]:
    y = q["values"].to_numpy()
    mc = InterpolatorFactory.create_1d_interpolator(t_nodes, y, InterpMethod.MONOTONE_CONVEX, ExtrapMethod.FLAT)
    node_err = max(abs(mc.interpolate(t) - v) for t, v in zip(t_nodes, y))
    mono_viol = conv_viol = 0.0
    for i in range(len(t_nodes) - 1):
        f = np.array([mc.interpolate(t) for t in np.linspace(t_nodes[i], t_nodes[i + 1], 501)])
        d1, d2 = np.diff(f), np.diff(f, 2)
        mono_viol = max(mono_viol, np.max(-np.sign(y[i + 1] - y[i]) * d1))
        conv_viol = max(conv_viol, np.max(-d2))
    print(f"{name}: max node error {node_err:.1e} | worst monotonicity violation {mono_viol:.1e} | worst convexity violation {conv_viol:.1e}")
```

**Method.** On each node interval $[x_i, x_{i+1}]$ of width $h$ with chord slope $s=(v_{i+1}-v_i)/h$, the IFR is a quadratic, so its slope is linear: $f'(t)=a+(b-a)t/h$. Hitting both end values forces $a+b=2s$, which leaves one free number, the left-end slope $a$.

**Monotone and convex.** The slope $f'$ goes from $a$ to $b=2s-a$ in a straight line.
- If $s>0$ pick $0\le a\le s$. Then $0\le a\le b$: the slope never changes sign (monotone increasing) and never decreases (convex).
- If $s<0$ pick $2s\le a\le s$. Then $a\le b\le 0$: monotone decreasing and convex.
- If $s=0$ the interval is flat.

That range for $a$ is never empty, so **every interval is handled whatever the data look like**, including non-monotone or concave sequences of nodes. Inside the range, $a$ is chosen to match harmonic-mean slope estimates at the two nodes (zero where the neighbouring chords change sign), so the curve is smooth wherever the data allow it. The price of guaranteeing the shape on every interval is that for data that are not globally convex the slope has a kink at a node (the curve stays continuous), which is the unavoidable cost: a $C^1$ curve cannot be convex on both sides of a node where the chord slope falls. Outside the nodes the IFR is held flat. The integral is exact (piecewise cubic antiderivative), so discount factors have no quadrature error.

### Why the PVs differ

The piecewise-constant (left-continuous) curve applies the *right-hand* node value to the whole interval before it. With rising IFRs that is the highest rate of the interval, so it overstates forwards everywhere except at the right node. The monotone-convex curve starts each interval at the lower left-node rate and rises to the right-node rate, so its forwards are lower on average and its discount factors are slightly larger.

- **FRA (receive fixed 3.2%)**: lower projected forward means a larger fixed-minus-float payoff, so the PV rises (+348.71).
- **RFR swap (receive fixed 3.4%, 3Y)**: the same effect accumulates over three annual floating periods on the steeper part of the curve, so the PV rises by the most (+6,618.32).
- **Overnight basis swap (receive SOFR + 10bp, pay Fed Funds)**: the SOFR minus FF gap widens from about 20bp to 65bp along the curve. Piecewise constant uses the wider right-node gap over each interval; the smooth curve uses a smaller average gap, so the SOFR leg gains less and the PV falls (-722.48).

The size of each difference tracks how much the curve moves inside the accrual window, not the product type.
