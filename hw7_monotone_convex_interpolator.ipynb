### Include an IFR curve plot and briefly explain how your method ensures monotonicity and convexity.

```python
import numpy as np
import matplotlib.pyplot as plt
from fixedincomelib.utilities.numerics import InterpolatorFactory, InterpMethod, ExtrapMethod

import QuantLib as ql

# Same time axis the yield curve uses: Actual/Actual (ISDA) year fraction from the value date.
def to_ql(d):
    ts = pd.Timestamp(d)
    return ql.Date(ts.day, ts.month, ts.year)
isda = ql.ActualActual(ql.ActualActual.ISDA)
t_nodes = np.array([isda.yearFraction(to_ql("2026-01-02"), to_ql(d)) for d in node_dates])
grid = np.linspace(t_nodes[0] - 0.3, t_nodes[-1] + 0.5, 800)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.2), sharex=True)
for ax, (name, q) in zip(axes, [("SOFR-1B", sofr_quotes), ("FF-1B", ff_quotes)]):
    y = q["values"].to_numpy()
    mc = InterpolatorFactory.create_1d_interpolator(t_nodes, y, InterpMethod.MONOTONE_CONVEX, ExtrapMethod.FLAT)
    pcp = InterpolatorFactory.create_1d_interpolator(t_nodes, y, InterpMethod.PIECEWISE_CONSTANT_LEFT_CONTINUOUS, ExtrapMethod.FLAT)
    ax.step(grid, [pcp.interpolate(t) for t in grid], where="post", color="0.65", lw=1.2, label="piecewise constant")
    ax.plot(grid, [mc.interpolate(t) for t in grid], color="C0", lw=2, label="monotone convex")
    ax.plot(t_nodes, y, "o", color="C3", label="IFR nodes")
    ax.set_title(f"{name} instantaneous forward rate")
    ax.set_xlabel("years from value date")
    ax.set_ylabel("IFR (annual decimal)")
    ax.grid(alpha=0.3)
axes[0].legend()
plt.tight_layout()
plt.show()

# Numerical check on every node interval: nodes are hit, the curve is monotone in the direction of the data, and it is convex.
for name, q in [("SOFR-1B", sofr_quotes), ("FF-1B", ff_quotes)]:
    y = q["values"].to_numpy()
    mc = InterpolatorFactory.create_1d_interpolator(t_nodes, y, InterpMethod.MONOTONE_CONVEX, ExtrapMethod.FLAT)
    node_err = max(abs(mc.interpolate(t) - v) for t, v in zip(t_nodes, y))
    mono_viol = conv_viol = 0.0
    for i in range(len(t_nodes) - 1):
        f = np.array([mc.interpolate(t) for t in np.linspace(t_nodes[i], t_nodes[i + 1], 501)])
        d1, d2 = np.diff(f), np.diff(f, 2)
        mono_viol = max(mono_viol, np.max(-np.sign(y[i + 1] - y[i]) * d1))
        conv_viol = max(conv_viol, np.max(-d2))
    print(f"{name}: max node error {node_err:.1e}; worst monotonicity violation {mono_viol:.1e}; worst convexity violation {conv_viol:.1e}")
```

**Choice of method.** The curve is built interval by interval from the supplied IFR points, in the spirit of Hagan and West: it is continuous, it passes through every node, and the shape conditions are enforced on each interval separately. A quadratic IFR on each interval is the simplest family that hits both end values and still leaves one free parameter to control the shape. It also keeps the integral, and therefore the discount factors, in closed form.

**Construction.** On each node interval $[x_i, x_{i+1}]$ of width $h$ with chord slope $s=(v_{i+1}-v_i)/h$, the IFR is a quadratic, so its slope is linear: $f'(t)=a+(b-a)t/h$. Hitting both end values forces $a+b=2s$, which leaves one free number, the left-end slope $a$.

**Monotonicity and convexity.** The slope $f'$ goes from $a$ to $b=2s-a$ in a straight line.
- If $s>0$, choose $0\le a\le s$. Then $0\le a\le b$, so the slope never changes sign (monotone increasing) and never decreases (convex).
- If $s<0$, choose $2s\le a\le s$. Then $a\le b\le 0$, so the curve is monotone decreasing and convex.
- If $s=0$, the interval is flat.

**Data that cannot satisfy all conditions.** The admissible range for $a$ is never empty, so every interval is handled whatever the data look like. Inside the range, $a$ is chosen to match harmonic-mean slope estimates at the two nodes (zero where the neighbouring chords change sign), so the curve is smooth wherever the data allow it. If the node values are not monotone, each interval is still monotone in the direction of its own chord and the curve changes direction at the node. If the chord slope falls from one interval to the next, the slope of the curve jumps down at the node. The curve stays continuous but has a kink there, which cannot be avoided: convexity on both sides of such a node requires $b_i\ge s_i>s_{i+1}\ge a_{i+1}$, so the slope cannot be continuous. Outside the nodes the IFR is held flat. The integral is exact (a piecewise cubic antiderivative), so discount factors carry no quadrature error.

### Why the PVs differ

The piecewise-constant (left-continuous) curve applies the right-hand node value to the whole interval before it. With rising IFRs that is the highest rate of the interval, so it overstates forward rates everywhere except at the right node. The monotone-convex curve starts each interval at the lower left-node rate and rises to the right-node rate, so its forward rates are lower on average and its discount factors are slightly higher.

- **FRA (receive fixed 3.2%).** A lower projected forward rate increases the fixed-minus-floating payoff, so the PV rises (+348.71).
- **RFR swap (receive fixed 3.4%, 3 years).** The same effect accumulates over three annual floating periods on the steeper part of the curve, so the PV rises by the most (+6,618.32).
- **Overnight basis swap (receive SOFR + 10 bp, pay Fed Funds).** The gap between the SOFR and Fed Funds IFRs widens from about 20 bp to 65 bp along the curve. Piecewise constant applies the wider right-node gap over each interval, while the smooth curve applies a smaller average gap, so the net floating amount received is lower and the PV falls (-722.48).

The size of each difference depends on how much the curve moves inside the accrual window, not on the product type.
